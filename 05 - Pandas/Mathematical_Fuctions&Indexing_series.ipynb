# Mathematicle Functions and Indexing in Series.

### Mathematicle Functions in Pandas

#### 1. sum() ---> Sum of all the values in the series.
#### 2. prod() ---> Product of all the values in the series.
#### 3. mean() ---> Mean provides us with the average value of the data.
#### ex. Company A has 5 employees with salaries of 100, 20, 40, 123, and 90. The mean salary is calculated as follows:
#### 100 + 20 + 40 + 123 + 90 = 373
#### Mean = 373 / 5 = 74.6
#### 4. median() ---> Median provides us with the middle value of the data.
#### median salary is calculated as follows:
#### 20, 40, 90, 100, 123
#### Median = 90
#### 5. mode() ---> Mode provides us with the most frequently occurring value in the data.
#### Mode is calculated as follows:
#### 20, 40, 20, 100, 213
#### Mode = 20
#### 6. Std() ---> Standard deviation provides us with the measure of how spread out the data is.
#### standard deviation is calculated as follows:
#### 20, 40, 90, 100, 123
#### Standard deviation = sqrt(((mode - mean)^2 + (mode - mean)^2 + (mode - mean)^2 + (mode - mean)^2 + (mode - mean)^2) / n)
#### Standard deviation = sqrt((20 - 74.6)^2 + (40 - 74.6)^2 + (90 - 74.6)^2 + (100 - 74.6)^2 + (123 - 74.6)^2) / 5
#### Standard deviation = sqrt( 4.6^2 + 6^2 + 45^2 + 100^2 + 42^2) / 5
#### Standard deviation = sqrt( 26.6^2) / 5
#### Standard deviation = 5.8
#### Low standard deviation ---> 7,8,7.5,6,6.5
#### High standard deviation ---> 8,15,132,23,42
#### 7. Var() ---> Variance provides us with the measure of how spread out the data is. it's the square of the standard deviation.
#### variance is calculated as follows:
#### variance = standard deviation^2
#### variance = 5.8^2
#### variance = 33.64
#### 8. sort_values() ---> Sorts the values in the series in ascending order.
#### 9. max() ---> Maximum value in the series.
#### 10. min() ---> Minimum value in the series.
#### 11. count() ---> Number of non-null values in the series.
#### 12. Describe() ---> It provides us with the idea about the summary of the important mathematical calculations for data

```python
import numpy as np
import pandas as pd
```

```python
Experiment = pd.Series(data = [18, 56, 58, 18, 2000], index = ['a', 'b', 'c', 'd', 'e'])
```

```python
Experiment.sum()
```

```python
Experiment.prod()
```

```python
Experiment.sort_values()
```

```python
Experiment.mean()
```

```python
Experiment.median()
```

```python
Experiment.mode()
```

```python
Experiment.std()
```

```python
Experiment.max()
```

```python
Experiment.min()
```

```python
Experiment.count()
```

```python
Experiment.var()
```

```python
Experiment.describe()
# count ---> number of non null values
# mean ---> average of the values
# std ---> standard deviation of the values
# min ---> minimum value of the values
# 25% ---> 25% of the values out of my all values are less than this specific value
# 50% ---> 50% of the values out of my all values are less than this specific value
# 75% ---> 75% of the values out of my all values are less than this specific value
# max ---> maximum value of the values
```
