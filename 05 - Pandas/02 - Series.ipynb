# Series

```python
import numpy as np
import pandas as pd
```

## Pandas series is a column of data in a table like in excel or Google sheets with label called as indexes attached to each value.

```python
pd.Series(data=[40,50,60], index=['Bread','Milk','Cheese'],name='Product Prices')
```

```python
# Series is a 1D array that can hold any type of data (integer, string, float, python objects, etc.).
```

#### Data Type ---> Integer ---> Data in Series(Column) ---> Quantity of items in Stock
#### Data Type ---> Float --->Data in a series(column) ---> Price or product ratings
#### Data Type ---> String --->Data in a series(column) ---> Product name
#### Data Type ---> Boolean --->Data in a series(column) ---> Is item is in stock
#### Data Type ---> Date and time --->Data in a series(column) ---> Expiry dates or sale date
#### Data Type ---> Mixed Type --->Data in a series(column) ---> milk/40/true

## Creating a series in multiple ways:
1. list
2. Numpy Array
3. Dictionary
4. Custom Index
5. Scaler Value

### Series from list

```python
Country = ['India', 'USA', 'China', 'Russia', 'Japan', 'Germany']
pd.Series(Country)
```

```python
batsman_runs = [100, 45, 78, 34, 56, 90]
pd.Series(batsman_runs, dtype = 'int32')
```

```python
subjects = ['Maths', 'Physics', 'Chemistry', 'Biology', 'English']
Marks = [98, 87, 75, 80, 95]
pd.Series(data=Marks, index=subjects, name='Student Marks', dtype='int32')
```

### Creating a series using numpy Array

```python
Values  = np.arange(10, 21)
pd.Series(Values)
```

### Dictionary to series
#### Dictionary is another data Strure that stores values in key-value pairs.
#### Just because it stores values in key-value pairs so we don't have to explicitly mention the index of the series. The keys of the dictionary will be used as the index of the series.

```python
data_items = {'Maths': 98, 'Physics': 87, 'Chemistry': 75, 'Biology': 80, 'English': 95}
pd.Series(data_items, name='Student Marks', dtype='int32')
```

### Custome index to Series

```python
values = [1, 2, 3, 4, 5]
pd.Series(data = values, index = ['a', 'b', 'c', 'd', 'e'])
```

### scaler value inside series

```python
pd.Series(4,index = ['a', 'b', 'c', 'd', 'e'])
```
