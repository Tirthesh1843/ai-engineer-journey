```python
# Convert temperature from celcius to feranheit
# feranheit to celcius

# if temperature in celcius, then conversion is to feranheit
'''1 deg celcius =  9/5 + 32 feranheit
45 deg celcius =  45*9/5 + 32 feranheit'''
# if temperature in feranheit, then conversion is to celcius1
'''1 fahrenheit = 5/9 - 32 celcius
45 fahrenheit = 5/9*(45-32) celcius'''
def temprature_changes (temp, unit):
    if temp.isdigit():
        temp = int(temp)
        unit = unit.upper()
        if unit == 'C':
            temperature = temp*9/5 + 32
            print(f'the temprature in feranheit is {temperature} for given {temp} celcius.')
        elif unit == 'F':
            temperature = 5/9*(temp-32)
            print(f'the temprature in celcius is {temperature} for given {temp} feranheit.')
        else:
            print('invalid unit')
    else:
        print('invalid temprature')
```

```python
temprature = input('Enter the temprature: ')
unit = input('Enter the unit if celcius than C, if feranheit then F: ')
temprature_changes(temprature, unit)
```

```python
# define a function that would behave as password strength checker.
'''
password length > 8.
Password should have at least one uppercase letter.
Password should have at least one lowercase letter.
Password should have at least one special character.
password should have at least one digit.
weak password: if password doesn't match any above condition than it's a weak password.
'''
```

### islower() --> is current character is lower or not.
### isupper() --> is current character upper or not.
### isdigit() --> checks for current character is digit or not.

```python
def password_checker(password):
    if len(password) < 8:
        print(f'The entered {password} is a weak password because it\'s length is less than 8.')
    elif not any (char.islower() for char in password):
        print(f'The entered {password} is a weak password because it doesn\'t contain any lowercase letter.')
    elif not any (char.isupper() for char in password):
        print(f'The entered {password} is a weak password because it doesn\'t contain any uppercase letter.')
    elif not any (char in ['@', '#', '$', '%', '&', '*'] for char in password):
        print(f'The entered {password} is a weak password because it doesn\'t contain any special character.')
    elif not any (char.isdigit() for char in password):
        print(f'The entered {password} is a weak password because it doesn\'t contain any digit.')
    else:
        print("The entered password is strong.")

```

```python
password = input('Enter the password: ')
password_checker(password)
```

## Palindrom string
### palindrom string is going to have characters similar to it from both sides or even if you will reverse the string still you will be getting the given string itself.

### palindrom -> aaa, aba, aabbaa, naman.

### not palindrom -> adcs, aaba

```python
def palindrom(string):
    string = string.lower().replace(' ', '')
    if string == string[::-1]:
        print(f'{string} is a palindrom.')
    else:
        print(f'{string} is not a palindrom.')
```

```python
palindrom('naman')
```

```python
palindrom("a man a plan a canal Panama")
```
