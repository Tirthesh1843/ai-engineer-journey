# OS Library

### OS Library is able to interect with our Operating System and can let us work with multiple Directories and files.

#### getcwd() ---> It is capable of providing us with the idea about the current working directory.
#### getcwd ---> get current working directory

```python
import os
```

```python
current_path = os.getcwd()
print("The current working directory is", current_path)
```

#### mkdir() ---> make directory ---> It will be helping us out with creating the directory.

```python
new_path = "NewFolder"
os.mkdir(new_path)
print(f"The new directory {new_path} is created successfully.")
```

#### listdir()---> List Directories ---> listing all of the directories present.

```python
print(os.listdir("."))
```

```python
## Joining two/multiple paths together.

# os.path.join
dir_name = "Folder_Data"
file_name = "File_data"
Full_path = os.path.join(dir_name, file_name)
print(f"The full path is: {Full_path}")
```

```python
## I want to join multiple path with the current path.
Current_complete_path = os.path.join(os.getcwd(), dir_name, file_name)
print(f"The complete path is: {Current_complete_path}")
```

```python
## os.path.exist()
given_path = 'requirements.txt'
if os.path.exists(given_path):
    print(f"The path {given_path} exists.")
else:
    print(f"The path {given_path} does not exist.")
```

```python
## os.path.isfile() ---> It checks that wether if current path is a file or not.
##os.path.isdir() ---> It checks that wether if current path is a directory or not.

path = 'NewFolder'
if os.path.isfile(path):
    print(f"The path {path} is a file.")
elif os.path.isdir(path):
    print(f"The path {path} is a Folder.")
else:
    print(f"The path {path} is not a file nor a Folder.")
```

```python
path = 'example.txt'
if os.path.isfile(path):
    print(f"The path {path} is a file.")
elif os.path.isdir(path):
    print(f"The path {path} is a Folder.")
else:
    print(f"The path {path} is not a file nor a Folder.")
```

```python
## The complete path <---> Absolute path ---> os.path.abspath()

## os.path.abspath() ---> It provides us with the complete path of the absolute path of the file or folder.
```

```python
file_name = 'example.txt' # Relative path
complete_path = os.path.abspath(file_name)
print(f"the complete path of {file_name} is: {complete_path}")
```
