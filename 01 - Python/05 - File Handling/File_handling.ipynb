## File Handling:

### 1. Read the content from my text file in system.

```python
## open() ---> opening your file.
## First parameter ---> File_name
## second parameter ---> mode
## mode ---> it's going to decide the way in which we are working with this file.
'''   Mode ---> r ---> reading from file
                w ---> Write content in file
                r+ ---> Read the content first then write content into the file
                w+ ---> Write the content first then reading from it.'''
```

```python
# Reading Your file.
## read() ---> It reads the contents from your file.
```

```python
f = open('example.txt', 'r')
content = f.read()
print(content)
f.close()
```

### Writing the content into our file.

##### write() ---> it overwrites the content into our text file.
mode ---> W content will be overwritten
##### append ---> It adds the data from the end of the file.
mode ---> a ---> It adds the data from the end of the file.

```python
# Write content in our file.
```

```python
f = open('example.txt','w')
f.write("Hello we are now writing the content into our file.")
f.close()
```

```python
# add the content without removing it using append.
f = open('example.txt','a')
f.write('\nWe are Learning about Gen AI.')
f.close()
```

## With ---> it will be used for first of all opening the file as object and there is no requirement for closing the file as well.

```python
with open ('example.txt','r') as f:
    content_data = f.read()
    print(content_data)
```

```python
with open('example.txt','w') as file:
    file.write('We are exploring about with condition in python file handling.')
```

```python
with open('example.txt','a') as files:
    files.write(' Hello, we are exploring append mode in python file handling.')
```

# File Handling questions.

```python
# Write the content into our file line by line ---> writelines()
```

```python
lines = ["First line \n", "Second line \n", "Third line \n"]
with open('example.txt','w') as data:
    data.writelines(lines)
```

```python
# r, w, a
```

```python
##   Modes ---> r ---> read
##              w ---> write
##              a ---> append
##              r+ ---> read and write
##              w+ ---> write and read
```

# Binary format files

```python
##   Modes ---> rb ---> read in binary format
##              wb ---> write in binary format
##              ab ---> append in binary format
##              r+b ---> read and write in binary format
##              w+b ---> write and read in binary format
```

```python
data = b'\x00\x01\x02\x03\x04'
with open('example.txt', 'wb') as binary_data:
    binary_data.write(data)
```

```python
with open('example.txt','rb') as binary_read:
    binary_content = binary_read.read()
    print(binary_content)
```

```python
# read the content from source.txt file and then write the content in another file.
```

```python
with open('Source.txt','r') as source:
    content = source.read()

with open('destination.txt','w') as destination:
    destination.write(content)
```

```python
## Reading the Total lines, Total Words, Total Characters from a file.
def count_lines(file):
    with open(file,'r') as f:
        lines = f.readlines()
        total_lines = len(lines)
        total_words=sum(len(line.split()) for line in lines)
        total_characters = sum(len(line) for line in lines)
    return total_lines, total_words, total_characters
file_path = 'Total_data.txt'
total_lines, total_words, total_characters = count_lines(file_path)
print(f"Total lines: {total_lines}",f"\nTotal words: {total_words}",f"\nTotal characters: {total_characters}")
```

```python
## Write the content into a file and then after writing the content into that file we are going to read the content from that file as well --> seek()
```

w+ ---> First of all as W is their so it states that first writing noperation than readin operation will be done.

seek() ---> It is going to help us out with reading the content from the starting.

```python
with open('Total_data.txt','w+') as file:
    file.write('\nWe are looking into W+ Operation.')
    file.write('\nIt means writing and then reading')
    file.seek(0)
    read_data = file.read()
    print(read_data)
```
