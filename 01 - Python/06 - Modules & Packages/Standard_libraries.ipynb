```python
### CSV Library ---> Most of the time we are going to work with csv files so that's why knowledge with csv library functionality matters. 
```

```python
import csv
```

```python
with open ('example.txt',mode='w',newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Name','Age'])
    writer.writerow(['Tirthesh',23])
```

```python
with open ('example.txt',mode='r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
## JSON Library:
## When we are going to work with Nosql Database data then the data inside the Nosql Database is going to be there in the JSON format and along with that when we are going to work with the Datasets as well than again this Dataset is going to be in the JSON format.
```

```python
## When we get JSON data, We get in a JSON string format so by using JSON Library we can convert our string data into Dictionary format for serializing our data.
```

```python
import json
```

```python
data = {'name':'Tirthesh','age':23,'gender':'male','qualification':'Btech','city':'Rajkot'}
str_data = json.dumps(data)
print(str_data)
print(type(str_data))
```

```python
json_dict = json.loads(str_data)
print(json_dict)
print(type(json_dict))
```

```python
## Regular Expression Library

## Regular Expression is used to retrieve info based on a specific pattern or if the pattern exist.
```

```python
import re
```

```python
text = 'Hello My name is Tirthesh Dhruv. I\'m Currently learning Python Basics. We are looking over on the functionality of Regular Expression library.also i\'m 23 years old.'
```

```python
pattern = r'\d+'
match = re.search(pattern,text)
print(match.group())
```
