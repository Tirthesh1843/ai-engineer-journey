# Map Function

### Map function applies a given function operation to all items in an iterable or a listand returns a map object (an Iterator).

### Map function is particularly useful for transforming the data or applying an operatrion on all items of the iterable.

### map() is going to help us out with avoiding unnecessary lines of code and we can apply the operation on all values, on all items in an iterable and this is the sign of a good programmer.

## Syntax for map():
###  map(function,iterable)

```python
## Square of all items in a list
def square(list_items):
    squares = []
    for item in list_items:
        squared_item = item ** 2
        squares.append(squared_item)
    print(squares)
```

```python
square([1,2,3,4,5])
```

```python
print(list(map(lambda x: x**2, [1,2,3,4,5])))
```

```python
# Map fuction with basic function in it.
def square_numbers(x):
    return x**2
print(list(map(square_numbers, [1,2,3,4,5,6,7])))
```

### Whenever we are going to pass the general function inside the map function as the parameter than we do not have to use parenthesis.

### While mapping the value of 1 iterable with another iterable we can actually apply apply the operation in between the values of the two iterable in a synchronous way.
#### [1,2,3,4] <---> [5,6,7,5] ---> Add ---> [6,8,10,9]

```python
# Map with multiple iterables
no_1 = [1,2,3,4,5]
no_2 = [6,7,8,9,10]
print(list(map(lambda x,y:x+y, no_1,no_2)))
```

```python
# Map() to convert all string values items in an iterable into an integer.

map_values = list(map(int,['1','2','3','4','5']))
```

```python
print (map_values)
```

```python
# convert all names to upper case
names = ['Tirth','Tirthesh','Priya','Karan','Premal']
Upper_names = list(map(str.upper,names))
print(Upper_names)
```

```python
## Retrieve the values inside a dictionary using map function

Employees = [{"name": "Tirthesh",'age':23,'Qualification':"B.E."},
            {"name": "Priya","age":23,'Qualification':"MBBS"},
            {"name": "Karan","age":24,"Qualification":"B.com."},]
```

```python
def age_retriver(x):
    return x["age"]
print(list(map(age_retriver,Employees)))
```
