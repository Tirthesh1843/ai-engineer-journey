# Recursive functions

## By a recursive function we can say that a function is going to call itself again and again.

```python
# Factorial of a number
# factorial of 8 ---> 8*7*6*5*4*3*2*1
# factorial of 98 ---> 98*97*96*...*3*2*1
# factorial of 0/1 ---> 1
```

```python
def factorial(n):
    if n == 0 or n == 1:
        print(f"Factorial of {n} is 1")
    else:
        factorial = 1
        for i in range (1, n+1):
            factorial = factorial*i
        print(f"Factorial of {n} is {factorial}")
```

```python
factorial(4)
```

```python
def recursive_factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * recursive_factorial(n-1)
```

```python
recursive_factorial(5)
```

## Function with Multiple parameters
#### In function with multiple parameters, We are going to pass multiple  values in our parameters or more than 1 parameter we can pass.

```python
# Addition of numbers
def addition(num1, num2):
    print(f"the addition of {num1} and {num2} is {num1+num2}")
```

```python
addition(10, 20) 
```

```python
addition(10, 20, 30)
```

## Function with variable length arguments:

### Positional arguments are being used so that we can pass multiple values inside our function as an argument in respective of the number of values that we have mantioned as an parameter inside our function.

```python
# function to add multiple numbers irrespective of how many numbers we are passing:

def addition_positional(*num):
    add = 0
    for i in num:
        add += i
    print("the addition of numbers are:", add)
    
```

```python
addition_positional(10, 20, 30, 40, 50,1,3,45,2,3)
```

### Keyword arguments takes the input as the keyword values and irrespective of any numbers of keyword values you are going to pass it won't matter.

```python
def data(**user_data):
    for key,value in user_data.items():
        print (f"{key} : {value}")
```

```python
data(name="Tirthesh", age=23, city="Chennai", state="Tamil Nadu", country="India", Gender = "Male")
```

### When we want to pass arguments with key value pairs along with positional arguments.

```python
def variable(*positional, **keyword):
    for i in positional:
        print(f"positional values are: {i}")
    for key,value in keyword.items():
        print(f"keyword values are: {key} : {value}")
```

```python
variable(1,2,3,4,5,name = "Tirthesh", age=23, Gender = "Male", 12)
```

```python
variable(1,2,3,4,5,name = "Tirthesh", age=23, Gender = "Male")
```
