# Lamda Functions

### Lamda functions are the one liner functions.It's defined by the lambda keyword.

## Syntax for lambda function:

#### lambda arguments: expression

#### lambd ---> Keyword which simply let the python interpreter know the current function is are lambda function
#### arguments ---> Input values to our function
#### Expression ---> Logic (function body)

##### lambda x:x**2

### They can have any no. of arguments but only 1 expression or one logic.
### They are used for sort operation or as arguments to higher order functions.

### They can be treated as function without a name and they do not even require any return statements.

```python
## Function which calculkates the square of numbers.

def square(n):
    return n**2
```

```python
print(square(2))
```

```python
# square using lambda

square_lambda = lambda n: n**2
print(square_lambda(2))
```

```python
print(type(square_lambda))
```

```python
add = lambda x, y: x + y
add(2, 3)
```
