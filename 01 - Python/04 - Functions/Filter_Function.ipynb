# Filter Function

### Filter function is going to retrieve only those items which are meeting certain condition in iterable.

### The filter function constructs an iterator from elements of an iterable for which a condition remains satisfy or for which function returns true value.

## Syntax for filter Function:

#### list(filter(func,iterable))

```python
# To retrieve only those values which are even.

def even_func(x):
    if x%2 == 0:
        return True
```

```python
iterable_list = [1,43,5,23,5,2,5,8,3,6,8,23,0,84,2,5]
print(list(filter(even_func, iterable_list)))
```

```python
#solving same with lambda:
print(list(filter(lambda x: x%2 == 0, iterable_list)))
```

```python
# retrieve the values which are greater than 5
print(list(filter(lambda x: x>5, iterable_list)))
```

```python
# greater than 5 and even
print(list(filter(lambda x: x>5 and x%2 == 0, iterable_list)))
```

```python
#iterating over a dictionary:
Employee = [{"Name":"Tirthesh", "Age": 23, "Qualification": "B.E."},
            {"Name":"Priya", "Age": 24, "Qualification": "MBBS"},
            {"Name":'Premal', "Age": 25, "Qualification": "Diploma"}]
```

```python
def age_greater(x):
    if x['Age'] < 25:
        return True
print(list(filter(age_greater, Employee)))
```
