### If we are getting an error and we are not aware that which specific error type it is then we can use the primary exception class which is Exception.

#### The parent exception type must be mentioned at the end or Last as it is the Parent class to only if the Previous exception class will not catch the error than this final one will catch the error.

#### try ---> try block is going to have the operation where we can except an error to occure.
#### except ---> except block is used to handle the error in a way such that the program execution shouldn't stop.
#### else ---> It will be consisting the code which we want to run if the try block operation is working without throwing error.
#### finally ---> finally block gets executed once we are done with everything and even if we are getting an error or not getting an error still the code inside the finally block gets executed.

#### if try ---> code is throwing an error ---> except ---> code will be running
#### if try ---> code is not throwing an Error ---> else ---> code will be running

```python
num = int(input("Enter the denominator for devision: "))
try:
    result = 1/num
except Exception as ex:
    print(ex)
else:
    print(f"the division value for {num} is {result}")
```
