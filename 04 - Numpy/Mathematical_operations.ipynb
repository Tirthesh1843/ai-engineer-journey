### axis ---> It provides us with the idea about the any operation value inside specific row or column or we can get a specific value row wise and column wise.

#### axis = 0 ---> It is used for column wise operations.
#### axis = 1 ---> It is used for row wise operations.

```python
np.max(array_data,axis=0)
```

```python
np.max(array_data,axis=1)
```
