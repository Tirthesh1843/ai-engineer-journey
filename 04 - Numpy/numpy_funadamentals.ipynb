## Numpy Arrays vs Python sequences:

#### Numpy arrays have a fixed size at creation, unlike python lists which grow dynamically.Changing the size of n dimentional array will create a new array and delete the original.

#### The elements in a Numpy array are required to be of the same Data Type and thus will be of the same size in memory.

#### Numpy arrays facilitate advanced Mathematicle and Other types of operations on Large number of data. Typically, Such operations are executed more efficiently with less code than is possible using Python's built in Sequences.
