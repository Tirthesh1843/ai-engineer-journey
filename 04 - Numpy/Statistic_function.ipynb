# Statistical Functions in Numpy

```python
## mean() ---> It will provide mean value of the array.
```

```python
import numpy as np
```

```python
array = np.round(np.random.random((4,4))*100).astype(int)
```

```python
array
```

```python
np.mean(array)
```

```python
np.mean(array, axis=0) ## It will provide mean value of the column.
```

```python
np.mean(array, axis=1) ## It will provide mean value of the row.
```

```python
## std() ---> It will provide standard deviation of the array.
```

```python
np.std(array)
```

```python
np.std(array, axis=0)
```

```python
np.std(array, axis=1)
```

```python
## median() ---> It will provide median value of the array.
```

```python
np.median(array)
```

```python
# axis=0 ---> It will provide  value of the column.
# axis=1 ---> It will provide value of the row.
```

```python
## var ---> Variance of all values in the matrix or array.
## syntax: np.var(array, axis=0)
```

```python
array
```

```python
np.var(array)
```

```python
# Dot product of two array/Matrices.
# Syntax: np.dot(matrix1, matrix2)
# rule: matrix1.shape[1] == matrix2.shape[0]
```

```python
matrix_1 = np.round(np.random.random((3,4))*100).astype(int)
matrix_2 = np.round(np.random.random((4,2))*100).astype(int)
```

```python
matrix_1
```

```python
matrix_2
```

```python
np.dot(matrix_1, matrix_2)
```

```python
np.dot(matrix_2, matrix_1)
```

## Indexing and slicing in numpy.

```python
## indexing --->list[,size-1] ---> To have the access over the values.
```

```python
array = np.round(np.random.random((3,4))*100).astype(int)
```

```python
array
```

```python
array_1 = np.arange(1,10)
```

```python
array_1
```

```python
array_1[6]
```

```python
array[2,3]
```

```python
## Indexing in 2D array ---> list[row_index, column_index]
```

```python
## Indexing in 3D array ---> list[depth_index, row_index, column_index]
```

```python
array3D = np.round(np.random.random((3,4,2))*100).astype(int)
```

```python
array3D
```

```python
array3D[1,1,1]
```

```python
array
```

```python
## slicing ---> list[start_index:end_index:step]
## 2d array ---> list[row_start_index:row_end_index:row_step, column_start_index:column_end_index:column_step]
```

```python
array[:,2] # column 3
```

```python
array[0,:] # row 1
```

```python
array[1:3,2:4]
```
