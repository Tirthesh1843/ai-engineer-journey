## stacking and splitting of the array

```python
## Stacking of arrays: (Size of the array should be same)
## hstack() ---> Horizontal Stack ---> Stacking with the same number of rows
## vstack() ---> Vertical Stack ---> Stacking with the same number of columns
## dstack() ---> Depth Stack ---> Stacking along a third axis
```

```python
array_one = np.arange(1,13).reshape(3,4)
array_two = np.arange(10,22).reshape(3,4)
```

```python
Horizontal_stack = np.hstack((array_one,array_two))
Horizontal_stack
```

```python
vertical_stack = np.vstack((array_one,array_two))
vertical_stack
```

```python
array3D_one = np.arange(1,13).reshape(2,2,3)
array3D_two = np.arange(10,22).reshape(2,2,3)
```

```python
array3D_one
```

```python
array3D_two
```

```python
Depth_stack = np.dstack((array3D_one,array3D_two))
Depth_stack
```

```python
# Splitting of arrays:
## hsplit() ---> Horizontal Split ---> Splitting by rows
## vsplit() ---> Vertical Split ---> Splitting by columns
## dsplit() ---> Depth Split ---> Splitting along a third axis
```

```python
array=np.arange(1,17).reshape(4,4)
```

```python
np.hsplit(array,2)
```

```python
np.vsplit(array,4)
```

```python
array=np.arange(1,19).reshape(3,3,2)
```
