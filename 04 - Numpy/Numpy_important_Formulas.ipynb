### reshape() : It helps us out with changing the shape of of an array as per our need.

```python
np.arange(2,17,2).reshape(4,2)
```

### ones() ---> Used for creating the array of ones.

```python
np.ones((3,3))
```

#### zeros() ---> Used for creating the array of zeros

```python
np.zeros((3,5), dtype=np.int32)
```

#### random() ---> It helps us out with generating any random values from 0 to 1 inside our matrix.

```python
np.random.random((3,4))
```

#### Identity() ---> It helps us out with creating the matrix as an Identity matrix. Rows and colums should be same.

```python
np.identity(3)
```

#### Linspace() ---> It makes sure that every value within the given range is going to have similar distance from each other.

```python
np.linspace(-10,10,10, dtype=int)
```
