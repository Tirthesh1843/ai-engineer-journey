```python
class Avenger:
    def __init__(self, name, power, wepon, nationality="America"):
        self.name = name
        self.power = power
        self.wepon = wepon
        self.nationality = nationality
    def intro(self):
        print(f"Hi I'm {self.name} and I'm from {self.nationality}. My power is {self.power}.")
    def carrying_wepon(self):
        print(f"I'm carrying {self.wepon}.")
```

```python
Ironman = Avenger("Tony Stark", "Genius of Technology", "Armorsuit")
Thor = Avenger("Thor", "Thunder", "Hammer","Asguard")
Hulk = Avenger("Bruce Banner", "Smash", "None")
Captain_America = Avenger("Steve Rogers", "Super Soldier", "Shield")
```

```python
Ironman.intro()
Ironman.carrying_wepon()
```

```python
Captain_America.intro()
Captain_America.carrying_wepon()
```

## Attributes ---> Attributes stands for the characteristics that a specific object is going to have.

## Methods ---> Functionality that is specific object is going to have.

## Self ---> self refers to the current object of the class.

## Types of Constructors:

### 1. Default constructor.
### 2. Parameterized constructor.

### 1. Default Constructor: It doesn't have parameters inside it and it will only have self keyword inside it. It assigns default values to all of the objects.

### 2. Parameterized Constructor: It consist of parameters inside it so that ever object can have specific attributes associated to it.
### It allows us to pass arguments while creating an object.

```python
# Default Constructor:
class Dog:
    def __init__(self):
        self.name = "Labrador"
        self.color = "Golden brown"
    def show_info(self):
        print(f"This dog is {self.name} and it's color is {self.color}.")
```

```python
dog1 = Dog()
```

```python
dog1.name
```

```python
dog1.show_info()
```

```python
dog2 = Dog()
dog2.name
```

```python
dog2.show_info()
```

```python
# parameterized constructor
class Dogs():
    def __init__(self,breed,color):
        self.breed = breed
        self.color = color
    
    def show_info(self):
        print(f"This dog is {self.breed} and it's having color as {self.color}.")
```

```python
dog1 = Dogs("Golden Retriever","Golden")
```

```python
dog1.breed
```

```python
dog1.show_info()
```

```python
dog_2 = Dogs("Husky","White")
```

```python
dog_2.show_info()
```

## Types of Methods:

### 1. Instance method : The instance method operate on object attributes.
#### The instance method is going to use self method to access and modify object attributes.
#### These methods are called using an object

### 2. Class method : It is the method for the whole class or the whole blueprint which will remain same for the all of the objects that such kind of method is reffered as class Method.
#### In class method, this methods work with the class level attributes.
#### They are going to use cls instead of self.
#### They are called using the class name as well.
#### For specific that the current method is a class method we have to use the abbreviation as @classmethod

### 3. Static method : Static Methods do not use self or cls.
#### They work like normal functions but belong to a class.
#### They are going to be designated by an abbreviation @staticmethod
#### Static Method can also be called by the class name also.

```python
class Avengers:
    def __init__(self,name,power):
        self.name = name
        self.power = power
    
    def introductio(self): #instance method
        print(f"Hi I'm {self.name} and my power is {self.power}.")
```

```python
iron_man = Avengers("Tony Stark","Genius of Technology")
```

```python
iron_man.introductio()
```

```python
captain = Avengers("Steve Rogers","Super Soldier")
captain.introductio()
```

```python
class Avengerss:
    origin = "Avenger"
    def __init__(self,name,power):
        self.name = name
        self.power = power
    @classmethod
    def origin_info(clf):
        print(f"My origin is {clf.origin}.")
    
    def self_intro(self):
        print(f"Hi I'm {self.name} and my power is {self.power}.")
```

```python
Ironman = Avengerss("Tony Stark","Billionair")
Ironman.power
```

```python
Ironman.self_intro()
```

```python
Ironman.origin_info()
```

```python
Captain_America = Avengerss("Steve Rogers","Super Soldier")
Captain_America.self_intro()
```

```python
Captain_America.origin_info()
```

```python
class Avengersss:
    @staticmethod
    def mission():
        print("Protect the world from evil.")

    origin = "Avenger"

    def __init__(self,name,power):
        self.name = name
        self.power = power
        
    @classmethod
    def origin_info(clf):
        print(f"My origin is {clf.origin}.")

    def introduction(self):
        print(f"Hi I'm {self.name} and my power is {self.power}.")
```

```python
spider_man = Avengersss("Peter Parker","Spider powers")
spider_man.mission()
spider_man.origin_info()
spider_man.introduction()
```

```python
iron_man = Avengersss("Tony Stark","Billionair")
iron_man.mission()
iron_man.origin_info()
iron_man.introduction()
```
