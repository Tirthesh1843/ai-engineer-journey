# Pillars of Object Oriented Programing

## Inheritance

#### Inheritance is a fundamental concept in object oriented programming because of which the child class is able to inherit the properties or attributes and functionalities of the parent class.

#### just like a child is going to inherit the properties of parent and parent is going to inherit the properties of grandparent and definitiely in this hierarchy the grand child can have some features from the parent and grandparent as well.

## In python typically we get to see 3 Types of Inheritance:
#### 1. Single Inheritance
#### 2. Multilevel Inheritance
#### 3. multiple Inheritance

### Single Inheritance ---> The features or the attributes from one single class will be getting inherited into one child class. 

#### Parent class (properties ---> Attributes and functionalies)
#### |
#### |
#### Child Class

```python
class Avenger:
    def __init__(self,name,power):
        self.name=name
        self.power=power
    
    def introduction(self):
        print(f"Hello my name is {self.name}")

class Ironman(Avenger):
    def __init__(self,name,power,wepon):
        super().__init__(name,power)
        self.wepon=wepon
    
    def wepon_details(self):
        print(f"my wepon is {self.wepon}")
```

```python
Iron = Ironman("Tony Stark","Genius of Technology","Armorsuit")
```

```python
Iron.name
```

```python
Iron.wepon
```

```python
Iron.wepon_details()
```

```python
Iron.introduction()
```

### Multilevel Inheritance ---> The properties or methods and attributes of grand parent class inherited by the parent class and then the properties of Parent class will be inherited by the child class. In a proper heirarchy the Inheritance process is going to work.

#### Grand parent (properties)
#### |
#### |
#### Parent (properties)
#### |
#### |
#### Child

```python
class Grandparent:

    def __init__(self,nationality, baldness, anger_issues):
        self.nationality = nationality
        self.baldness = baldness
        self.anger_issues = anger_issues

    def introduction(self):
        print(f"Hello my nationality is: {self.nationality}. \nAbout my anger issue if you want to know: {self.anger_issues}")
```

```python
class Parent(Grandparent):

    def __init__(self,nationality,baldness,anger_issues, intelligence):
        self.intelligence = intelligence
        super().__init__(nationality,baldness,anger_issues)
    
    def intelligence_details(self):
        print(f"my intelligence is {self.intelligence}")
```

```python
class Child(Parent):
    def __init__(self,nationality, baldness, anger_issues, intelligence, charm):
        super().__init__(nationality,baldness,anger_issues, intelligence)
        self.charm = charm

    def charm_details(self):
        print(f"wether you want to know if i have charm or not {self.charm}.")
```

```python
parent_1 = Parent("Indian", "Yes", "No", "Medium")
```

```python
parent_1.introduction()
```

```python
child_1 = Child("Indian", "Yes", "No", "Medium", "Yes")
```

```python
child_1.charm_details()
```

```python
child_1.intelligence_details()
```

```python
child_1.nationality = "America"
```

```python
child_1.nationality
```

## Multiple Inheritance

### In multiple Inheritance, The properties of two parent class or multiple parent class will be getting inherited into child class.

#### Parent_1(properties)+ Parent_2(Properties) + Parent_3(Properties + ...)
#### |
#### |
#### Child

```python
class Ironmans:

    def __init__(self,name,power,suit,wepon,nationality="American"):

        self.name=name
        self.power=power
        self.suit=suit
        self.wepon=wepon
        self.nationality=nationality
    
    def introduction(self):
        print(f"Hello my name is {self.name} and my nationality is {self.nationality}.")

class Captains:
    def __init__(self,name,power,suit,wepon,nationality="American"):

        self.name=name
        self.power=power
        self.suit=suit
        self.wepon=wepon
        self.nationality=nationality
    
    def wepon_details(self):
        print(f"my wepon is {self.wepon}.")

class Hulk(Ironmans,Captains):

    def __init__(self,name,power,suit,wepon,nationality,anger):
        super().__init__(name,power,suit,wepon,nationality)
        self.anger=anger

    def anger_issue(self):
        print(f"my anger issue is {self.anger}.")
```

```python
hulk_guy = Hulk("Bruce Banner","Power and strength","Hulk Buster","Hands and legs","American","Too much anger")
```

```python
hulk_guy.introduction()
```

```python
hulk_guy.wepon_details()
```

```python
hulk_guy.anger_issue()
```
