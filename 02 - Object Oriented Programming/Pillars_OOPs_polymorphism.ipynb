# Polymorphism

## Polymorphism means MANY FORMS in Greek.
## In OOPs, Polymorphism allows objects of different classes as if thery are the object of the same class and it enables reusability and makes programs more flexible.

## Types of Polymorphism:
1. Method Overriding (Runtime polymorphism)
2. Method Overloading (Compile Time Polymorphism)
3. Operator Overloading

### Method Overriding ---> A child class provides a specific implementation of a method that has already been defined into the parent class. This allows dynamic selection at Runtime.

```python
class Grandparent: # Parent class
    def speak(self):
        print("Hello My name is Tirth.")

class Parent(Grandparent): # Child class
    def speak(self):
        print("Hello My name is Jr. Tirth.")

class Child(Grandparent): # Child class
    def speak(self):
        print("Hello My name is Jr. Tirth junior.")
```

```python
def attribute_provider(objects):
    return objects.speak()
```

```python
grandpa = Grandparent()
```

```python
attribute_provider(grandpa)
```

```python
parent = Parent()
```

```python
attribute_provider(parent)
```

```python
child = Child()
```

```python
attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "This method returns the area of the Shape."
```

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
```

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius
```

```python
shape = Shape()
```

```python
print(shape.area())
```

```python
rectangle = Rectangle(10, 20)
```

```python
print(rectangle.area())
```

```python
circle = Circle(5)
```

```python
print(circle.area())
```
