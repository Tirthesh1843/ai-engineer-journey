# Magic Methods

## Magic methods are also called as dunder Methods because they have double underscores (__) before and after their name. And they are special methods in Python that allows us to customize how object behave.

## This Methods can help us define:
### 1. Object Initialization (__init__)
### 2. String representation (__str__,__repr__)
### 3. Arithmetic Operations (__add__,__sub__, etc)
### 4. Comparison Operations (__eq__,__lt__)
### 5. Length, Indexing, Iteration (__len__,__getitem__,etc)

## Why to use Magic Methods?

### 1. It makes objects behave like built in types.
### 2. It improves Redabilitive by allowing basic operation (object1 + object2)
### 3. Simplifies code by replacing verbose function calls with natural syntax.

###  print (obj) --> __str__ --> What is returned when print (object) is used.

### Arithmetic Operations can be done among the objects by using Magic Methods.

### __add__

```python
class Avenger:
    def __init__(self,name,strength):
        self.name = name
        self.strength = strength
    
    def __add__ (self,other):
        return Avenger(self.name +" & "+ other.name, self.strength + other.strength)

    def __str__ (self):
        return f"{self.name} has {self.strength} strength"

ironman = Avenger("Tony Stark", 123)
thor = Avenger("Thor", 100)
team = ironman + thor
print(team)
```
