# Operator Overloading

## Operator Overloading it allows us to define how Operators (+,-,*,/,etc) work with user defined objects.

## Why is Operator Overloading useful?
### Makes custom objects behave like built in types.(numbers,strings,list)
### Improves code redability by using natural syntax (object1 + object2) instead of object1.add(object2)

### operator ---> + ---> Method name ---> __add__(self,other)
### Operator ---> - ---> Method name ---> __sub__(self,other)
### Operator ---> * ---> Method name ---> __mul__(self,other)
### Operator ---> / ---> Method name ---> __truediv__(self,other)
### Operator ---> == ---> Method Name ---> __eq__(self,other)
### Operator ---> > --->Method name ---> __gt__(self,other)

```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power = power

    def __add__(self,other):
        return Avenger(self.name +" & " + other.name, self.power + other.power)
    
    def __sub__(self,other):
        return Avenger (self.name. self.power-other.power)
    
    def __mul__ (self,other):
        return Avenger (self.name, self.power*other.power)
    
    def __truediv__(self,other):
        return Avenger (self.name, self.power/other.power)

    def __str__(self):
        return f"name : {self.name}, power : {self.power}"

    def __gt__(self,other):
        return self.power > other.power
```

```python
thor = Avenger("Thor", 100)
spiderman = Avenger("Spiderman", 92)
```

```python
team = thor + spiderman
print(team)
```

```python
team2 = thor>spiderman
print(team2)
```
