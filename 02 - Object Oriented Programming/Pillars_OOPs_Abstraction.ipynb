# Abstraction

## Abstraction is the concept of hiding the implementation details and showing the only necessary features of object.
## This helps in reducing programming complexity and efforts.

### Abstraction allows us to work with an object with what an object does rather than how it does it.

## How we achieve abstraction in Python?

#### 1. Absract class ---> A class that cannot be instanciated that means for an abstract class we can't create the object of this class.

#### 2. Abstract Method ---> Abstract Method just because it can't be accessed by the object of the abstract class itself that's why we have to call this abstract method in our child class. And the implementation of the abstract method is not provided in the abstract class but must be provided in the child class.

#### The abstract method is being defind by 
### @abstractmethod

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car starts with an Ignition key.")

    def stop(self):
        print("Car stopped with break pads.")

class ElectricScooter(Vehicle):

    def start(self):
        print("Electric scooter starts with an power button.")

    def stop(self):
        print("Electric scooter stopped with hand break.")
```

```python
car = Car()
```

```python
car.start()
```

```python
electric_scooter = ElectricScooter()
```

```python
electric_scooter.start()
```
