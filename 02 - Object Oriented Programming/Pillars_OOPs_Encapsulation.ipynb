# Encapsulation

## Encapsulation is a way of wrapping data (Variables and Methods) inside a capsule or a class as a single unit.

### It restricts the direct access to some of the object components which is a means o preventing accidental inference and misuse of the data.

## Key Concepts of Encapsulation:

### 1. Public Members
### 2. Protected Members
### 3. Private Members
### 4. Getter and setter Methods

## 1. Public Member: 
### Accesible from anywhere or attributes are accesible from outside of the class as well.
### This Members are accesible by the derived class as well.
### Public member attribute is directly declared.

```python
class Ironman:
    def __init__(self,name,power,wepon):
        self.name = name
        self.power = power
        self.wepon = wepon
    def introduction(self):
        print(f"Hello my name is {self.name}. & My Power is {self.power} & I use {self.wepon}.")
```

```python
## dir() ---> It takes the input as the object name and gives us the idea that which member is public or protected or private.
```

```python
iron = Ironman("Tony Stark","Genius of Technology","Armorsuit")
```

```python
dir(iron)
```

```python
iron.introduction()
```

```python
iron.name
```

```python
iron.power
```

## 2. Private Member:
### Private member is not accesible outside of the class and not even in the derived class,
### In order to have the access over this private members or to bring the modification over this  Members we use getter and setter methods.
### Private members are declared by using "__" before the name

```python
class Avenger:

    def __init__(self,name,phone_number,Power):
        self.name = name
        self.__phone_number = phone_number
        self.power = Power
```

```python
aveng = Avenger("Tony Stark","1234567890","Genius of Technology")
```

```python
aveng.name
```

```python
aveng.phone_number
```

```python
dir(aveng)
```

```python
## Getter Function
```

```python
## getter method is going to help us out with accessing the private data.
```

```python
class Captain:
    def __init__(self,name,address):
        self.name = name
        self.__address = address
    ## getter method
    def access_address(self):
        print(f"My address is {self.__address}.")
```

```python
captain = Captain("Steve Rogers","New York")
```

```python
captain.address
```

```python
dir(captain)
```

```python
captain.__address
```

```python
captain.access_address()
```

```python
## setter method ---> Setter method is being used for modifying the private data.
```

```python
class Animal:
    def __init__(self,name,sound):
        self.name = name
        self.__sound = sound
    
    def sound_info(self): # getter method
        print(f"The sound of {self.name} is {self.__sound}.")
    
    def set_sound(self,new_sound): # setter method
        self.__sound = new_sound
        print(f"The new sound of {self.name} is {self.__sound}.")
```

```python
animal = Animal("Dog","Bark")
```

```python
animal.name
```

```python
animal.sound
```

```python
animal.sound_info()
```

```python
animal.set_sound("Meow")
```

```python
animal.sound_info()
```

## 3. Protected Data:
### It's the data which is not accesible outside of the class But the protected data is going to be accesible by derived class or child class.
### For having the protected member you canmantion it with _{attribute_name}.

```python
class Grandparent:
    def __init__(self,name,locker_key):
        self.name = name
        self._locker_key = locker_key
    
    #getter method
    def pass_key_info(self):
        print(f"The locker key information is  {self._locker_key}.")
```

```python
class Parent(Grandparent):
    def __init__(self,name,locker_key,):
        super().__init__(name,locker_key)
    
    def child_reveal(self):
        print(f"The locker key information is {self._locker_key}.")
```

```python
grandpa = Grandparent("Bipin","1234567890")
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
dir(grandpa)
```

```python
grandpa.pass_key_info()
```

```python
parent = Parent("Mehul","1234567890")
```

```python
parent.child_reveal()
```
